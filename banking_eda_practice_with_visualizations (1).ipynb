## 5) Clean text columns using `.str.strip()`
Some categories have extra spaces; strip them.

```python
obj_cols = df.select_dtypes(include="object").columns #string
obj_cols
```

```python
# Select text columns
obj_cols = df.select_dtypes(include="object").columns

# Remove extra spaces from each text column
for column in obj_cols:
    df[column] = df[column].astype(str)
    df[column] = df[column].str.strip()

# Display the first five rows
df[obj_cols].head()
```

```python
df[obj_cols] = df[obj_cols].apply(lambda s: s.astype("string").str.strip())
df[obj_cols].head()
```

## 6) Data type conversion (very important)
### A) Convert strings with commas → numeric using `to_numeric`
Some values look like `'110,000'`.

### B) Convert date string → datetime using `to_datetime`

```python
df["MonthlyIncome"]
```

```python
# Remove commas then convert to numeric
df["MonthlyIncome"] = pd.to_numeric(df["MonthlyIncome"].astype("string").str.replace(",", ""), errors="coerce")
df["Balance"] = pd.to_numeric(df["Balance"].astype("string").str.replace(",", ""), errors="coerce")
#abc,missing - nan
# Convert date string to datetime
df["AccountOpenDate"] = pd.to_datetime(df["AccountOpenDate"], errors="coerce")

df.dtypes
```

## 7) Fill missing values (basic)
- Numeric → median
- Categorical → mode

```python
num_cols = df.select_dtypes(include=["int64","float64"]).columns
cat_cols = df.select_dtypes(include="object").columns

df[num_cols] = df[num_cols].fillna(df[num_cols].median(numeric_only=True))
df[cat_cols] = df[cat_cols].fillna(df[cat_cols].mode().iloc[0])

df.isnull().sum().sort_values(ascending=False).head(10)
```

## 12) `where()` example
Mark unrealistically negative income as missing, then fill again.

```python
df["MonthlyIncome"] = df["MonthlyIncome"].where(df["MonthlyIncome"] > 0, np.nan)
df["MonthlyIncome"] = df["MonthlyIncome"].fillna(df["MonthlyIncome"].median())
df[["MonthlyIncome"]].head()
```

## 13) Groupby + aggregation

```python
(df.groupby("Segment")
   .agg(avg_balance=("Balance","mean"),
        avg_income=("MonthlyIncome","mean"),
        customers=("CustomerID","count"),
        churn_rate=("Churned", lambda s: (s=="Yes").mean()*100))
   .sort_values("avg_balance", ascending=False)
).head(10)
```

## 17) Binning with `pd.cut()`
Create age groups.

```python
bins = [0, 25, 35, 45, 60, 100]
labels = ["<=25","26-35","36-45","46-60","60+"]
df["AgeGroup"] = pd.cut(df["Age"], bins=bins, labels=labels)
df["AgeGroup"].value_counts()
```

## 18) Replace values (basic cleaning)

```python
# Example: replace 'unknown' education with 'secondary'
df["Education"] = df["Education"].replace("unknown", "secondary")
df["Education"].value_counts().head()
```

# Visualizations (Matplotlib)
We keep visuals simple for classroom demo.

## 19) Histogram: Balance distribution

```python
plt.figure()
plt.hist(df["Balance"], bins=30)
plt.xlabel("Balance")
plt.ylabel("Frequency")
plt.title("Balance Distribution")
plt.show()
```

## 20) Bar chart: Job counts

```python
plt.figure()
job_counts = df["Job"].value_counts().head(8)
plt.bar(job_counts.index.astype(str), job_counts.values)
plt.xticks(rotation=45)
plt.xlabel("Job")
plt.ylabel("Count")
plt.title("Top Job Categories")
plt.show()
```

## 21) Boxplot: Balance by Segment

```python
plt.figure()
# Prepare data for boxplot
segments = df["Segment"].unique()
data = [df.loc[df["Segment"]==seg, "Balance"] for seg in segments]
plt.boxplot(data, labels=[str(s) for s in segments], showfliers=False)
plt.xticks(rotation=45)
plt.xlabel("Segment")
plt.ylabel("Balance")
plt.title("Balance by Segment (Boxplot)")
plt.show()
```

## 22) Scatter: MonthlyIncome vs Balance

```python
plt.figure()
plt.scatter(df["MonthlyIncome"], df["Balance"], s=8)
plt.xlabel("Monthly Income")
plt.ylabel("Balance")
plt.title("Income vs Balance (Scatter)")
plt.show()
```

## 23) Line plot: Customers opened accounts over time

```python
opened_per_month = df.set_index("AccountOpenDate").resample("M")["CustomerID"].count()
plt.figure()
plt.plot(opened_per_month.index, opened_per_month.values)
plt.xlabel("Month")
plt.ylabel("New Accounts")
plt.title("Accounts Opened Per Month")
plt.show()
```
