```python
df = df.drop_duplicates()
```

```python
#Clean text columns using .str.strip()
df['job'] = df['job'].str.strip()
df['marital'] = df['marital'].str.strip()
df['education'] = df['education'].str.strip()
df['default'] = df['default'].str.strip()
df['housing'] = df['housing'].str.strip()
df['loan'] = df['loan'].str.strip()
df['contact'] = df['contact'].str.strip()
df['month'] = df['month'].str.strip()
df['poutcome'] = df['poutcome'].str.strip()
```

```python
obj_cols = df.select_dtypes(include='object').columns
obj_cols
```

```python
df["Job"]
```

```python
for column in obj_cols:
  df[column]= df[column].astype(str)
  df[column]= df[column].str.strip()
df[obj_cols].head()
```

```python
df[obj_cols] = df[obj_cols].apply(lambda x: x.astype(str).str.strip())
df[obj_cols].head()
```

```python
#data type conversion
```

```python
df["MonthlyIncome"]
```

```python
df["MonthlyIncome"] = df["MonthlyIncome"].astype(str)
```

```python
df["MonthlyIncome"] = df["MonthlyIncome"].str.replace(",","")
```

```python
df["MonthlyIncome"] = pd.to_numeric(df["MonthlyIncome"], errors="coerce")
```

```python
df["MonthlyIncome"] = pd.to_numeric(df["MonthlyIncome"].astype(str).str.replace(",",""), errors="coerce")#"raise","ignore", missing - NaN
```

```python
df["Balance"] = pd.to_numeric(df["Balance"].astype(str).str.replace(",",""), errors="coerce")#"raise","ignore", missing - NaN
```

```python
df["AccountOpenDate"] = pd.to_datetime(df["AccountOpenDate"], errors="coerce")
```
